## **Step 2: Create applicants.csv**

```python
from faker import Faker
import pandas as pd
import random

fake = Faker()

applicants = []

universities = [
    "University of Nairobi",
    "Kenyatta University",
    "Moi University",
    "Egerton University",
    "JKUAT University",
    "Maseno University",
    ""
]

statuses = ["Under Review", "Selected", "Rejected", "SELECTED", "selected", "UNDER REVIEW", ""]
domain = ["web dev", "Web development", "webdev", "WEB DEV", "web development",
          " web development ", "Applied Statistics", "APPLIED Statistics", "Data SCIENCE", "COMPUTER sci", "PURE MATHEMATICS"]
date = ["13/05/2026", "2026-05-12"]

for i in range(100):
    applicants.append({
        #Applicants with missing applicant names
        "Applicant_Name":(
            None if i >= 5 and i < 8 #Created missing values between line number index 5 and line index 8 excluding the line index 8 .
            else fake.name()
        ),
        #Applicants with missing emails
        "Email" : (
            None if i >= 98 #Created two missing values
            else fake.email()
            
        ),
        "Phone": fake.phone_number(),
        "Domain_Applied":random.choice(domain),
        "University_Name": (
            None if i == 21 #Creating a missing value in line index 21
            else random.choice(universities)
        ),
        "Application_date": random.choice(date),
        "Status": random.choice(statuses)
        
    })

df = pd.DataFrame(applicants)

df.to_csv("applicants.csv", index=False)

#If you have many rows, pandas may hide some of them. To display everything, use:
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Duplicate the first three rows
df = pd.concat([df, df.iloc[:3]], ignore_index=True)

df
```

## **Step 4: Handle Blanks and Missing Values (Column by Column, With Justifiaction)**

```python
#Removing blanks
df["University_Name"] = df["University_Name"] .replace("", "Not provided")
df["Status"] = df["Status"] .replace("", "Under Review")
```

This section removes both blanks and missing values in the University_Name and Status columns

```python
#Confirm if the University_Name blanks have been removed
(df["University_Name"] == "").sum().sum()
```

```python
#Confirm if the Statuses blanks have been removed
(df["Status"] == "").sum().sum()
```

```python
#Applicant_Name Missing?This is critical identifying information- a row without a name is not useful.
#Consider dropping these rows.
df = df.dropna(subset = ['Applicant_Name'])
```

A row without an applicant name is not usefull. So I dropped the rows with missing Applicant_name.

```python
#Email missing?Also fairly critical
df['Email'] = df['Email'].fillna('Not_provided@gmail.com')
```

I filled the missing email with "Not_provided@gmail.com". This is because the row contains other important vaues required in data analysis.

```python
#University_Name Missing?
df['University_Name'] = df['University_Name'].fillna('Not Specified')
```

I filled missing University_Name with "Not Specified" instead of dropping, since it is not essentaial in determing application status.

## **Step 6: Standardize Text Fields**

### (i) Standardizing the Domain_Applied Column

```python
df['Domain_Applied'].unique()# To confirm the domains in the original dataset
```

There were 11 Domain_Applied before standardization 

```python
domain_mapping = {
    'web dev' : 'Web Development',
    'webdev' : 'Web Development',
    'web development' : 'Web Development',
    'WEB DEV' : 'Web Development',
    'Web development' : 'Web Development', 
    ' web development ' : 'Web Development', #Removing the white space
    'Web development' : 'Web Development',
    'Applied Statistics' : 'Applied Statistics',
    'APPLIED Statistics' : 'Applied Statistics',
    'Data SCIENCE' : 'Data Science',
    'COMPUTER sci' : 'Computer Science',
    'PURE MATHEMATICS' : 'Pure Mathematics',
}
df['Domain_Applied'] = df['Domain_Applied'].map(domain_mapping)
```

```python
df['Domain_Applied'].unique()# To confirm the domains in the dataset after standardization
```

There were 5 domains applied after standardization, that is 'Web Development', Pure Mathematics, Computer Science and 'Applied Statistics' and Data Science.

### (ii) Standardizing the Status Column

```python
df['Status'].unique()#Checking the contents of the status column
```

The Status Column comprised 6 inconsistent values: 'selected', 'Selected', 'SELECTED', 'Under Review', 'UNDER REVIEW' and 'Rejected'.

```python
#Ensure that all the values in Status column become one consistent value. 
df['Status'] = df['Status'].str.strip().str.title()
```

```python
df['Status'].unique() #Checking the values in the Status column are consistent
```

The standardized Status column has 3 consistent values: 'Selected', 'Under Review' and 'Rejected'. 



## **Step 7: Fix Data Types (Especially Dates and Phone Numbers)**

```python
#Fixing the date column
df['Application_date'] = pd.to_datetime(df['Application_date'], errors = 'coerce')
```

```python
df['Application_date'].isnull().sum()
```

This created 46 missing dates

```python
#Fixing this
#Convert all the valid dates into Pandas datime format
df["Application_date"] = pd.to_datetime(
    df["Application_date"],
    format = "mixed", #This al;ows Pandas to recognise different formats
    dayfirst = True,
    errors = "coerce" # Converts invalid dates into missing values
)
```

I replaced the missing dates 2026-05-13

```python
df["Application_date"] = df["Application_date"].fillna(pd.to_datetime("2026-05-13"))
```

I displayed the dates consistently as YYYY=MM-DD

```python
#Fixing the Phone Column
```

I removed dashes and spaces to create a consistent format

```python
df['Phone'] = df['Phone'].str.replace("-", '', regex = False).str.replace(' ', '', regex = False)
```
